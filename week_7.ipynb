## Problem 1


Problem 1: In this problem, you are asked to complete a hand-written digit recognition task for MNIST dataset (i.e. multi-class classification task with 10 classes). In this project, you are asked to design a fully-connected neural network (FCNN).

Hints:

1. The information of MNIST dataset can be found in https://www.tensorflow.org/datasets/catalog/mnist;
2. The MNIST dataset can be directly downloaded via:


```python 

(train_images, train_labels), (test_images, test_labels) = datasets.mnist.load_data()

```


3. Dense layer can be used for realizing the FCNN;
4. Since this problem can be modeled as multi-class classification problem with 10 different
classes, the output layer needs to have 10 computing units.

````python
import numpy as np
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# 1. Load MNIST
(train_images, train_labels), (test_images, test_labels) = datasets.mnist.load_data()

print(train_images.shape)  # (60000, 28, 28)
print(test_images.shape)   # (10000, 28, 28)
````

````python
# Preprocessing of the data before feeding to the Neural Network

# 2. Normalize to [0, 1]
train_images = train_images.astype("float32") / 255.0
test_images  = test_images.astype("float32") / 255.0

# 3. Flatten: (N, 28, 28) -> (N, 784)
train_images = train_images.reshape(-1, 28 * 28)
test_images  = test_images.reshape(-1, 28 * 28)

num_classes = 10
````

````python
# The first model consist of classifc fully-connected neural network with 2 hidden layers, and 2 regularization drop-out layers

model = models.Sequential([
    layers.Input(shape=(28 * 28,)),         # Input layer (flattened image)
    layers.Dense(256, activation='relu'),   # Hidden layer 1
    layers.Dropout(0.2),                    # Regularization
    layers.Dense(128, activation='relu'),   # Hidden layer 2
    layers.Dropout(0.2),
    layers.Dense(num_classes, activation='softmax')  # Output layer
])

# The second model consist of classifc fully-connected neural network with 4 hidden layers, to check if by increasing depht we can improve performance

model_2 = models.Sequential([
    layers.Input(shape=(28 * 28,)),         # Input layer (flattened image)
    layers.Dense(256, activation='relu'),   # Hidden layer 1
    layers.Dropout(0.2),                    # Regularization
    layers.Dense(256, activation='relu'),   # Hidden layer 2
    layers.Dropout(0.2),                    # Regularization
    layers.Dense(128, activation='relu'),   # Hidden layer 3
    layers.Dropout(0.2),                    # Regularization
    layers.Dense(128, activation='relu'),   # Hidden layer 4
    layers.Dropout(0.2),
    layers.Dense(num_classes, activation='softmax')  # Output layer
])
````

````python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
````

````python
history = model.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=2
)
````

````python
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=0)
print(f"Test accuracy: {test_acc:.4f}")
````

````python
pred_probs = model.predict(test_images)
pred_labels = np.argmax(pred_probs, axis=1)

print("First 10 predictions: ", pred_labels[:10])
print("True labels:          ", test_labels[:10])
````

````python
model_2.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model_2.summary()
````

````python
history = model_2.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=2
)
````

````python
test_loss, test_acc = model_2.evaluate(test_images, test_labels, verbose=0)
print(f"Test accuracy: {test_acc:.4f}")
````

````python
pred_probs = model_2.predict(test_images)
pred_labels = np.argmax(pred_probs, axis=1)

print("First 10 predictions: ", pred_labels[:10])
print("True labels:          ", test_labels[:10])
````

As we see, by increasing more layers, we didnt improve the performance, which actually was quite good with the first iteration. Also may be a more sophisticate arquitecture could improve performance, but we will check on that on the next problem.

## Problem 2

Problem 2: In this problem, you will focus on the same hand-written digit recognition task for MNIST dataset. In this project, you are asked to design a convolutional neural network (CNN) with two convolutional layers and two max-pooling layers.

Hints:
1. Realization of convolutional layer can be found in: https://www.tensorflow.org/api_docs/python/tf/keras/layers/Conv2D;
2. Realization of max-pooling layer can be found in: https://www.tensorflow.org/api_docs/python/tf/compat/v1/layers/MaxPooling2D;
3. Realization of flatten layer can be found in: https://www.tensorflow.org/api_docs/python/tf/keras/layers/Flatten;

````python
# 1. Load data
(train_images, train_labels), (test_images, test_labels) = datasets.mnist.load_data()

# 2. Normalize to [0, 1]
train_images = train_images.astype("float32") / 255.0
test_images  = test_images.astype("float32") / 255.0

# 3. Add channel dimension: (N, 28, 28) -> (N, 28, 28, 1)
train_images = np.expand_dims(train_images, axis=-1)
test_images  = np.expand_dims(test_images, axis=-1)

num_classes = 10
````

````python
model = models.Sequential([
    # First conv block
    layers.Conv2D(filters=32,
                  kernel_size=(3, 3),
                  activation='relu',
                  input_shape=(28, 28, 1)),
    layers.MaxPooling2D(pool_size=(2, 2)),

    # Second conv block
    layers.Conv2D(filters=64,
                  kernel_size=(3, 3),
                  activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),

    # Flatten + fully connected layers
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(num_classes, activation='softmax')   # 10 classes (0–9)
])
````

````python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
````

````python
history = model.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=2
)
````

````python
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=0)
print(f"Test accuracy: {test_acc:.4f}")
````

## Problem 3

Problem 3: The IMDB dataset is a set of 50,000 highly polarized reviews from the Internet Movie Database. In this problem, your task is to develop an LSTM-based recurrent neural network model to automatically identify whether a given review is positive or negative.

Hints:
1. The information of the IMDB dataset can be found in https://www.tensorflow.org/api_docs/python/tf/keras/datasets/imdb.

By looking at the website, we see that the dataset is already tokenize, so we only need to set a couple of hyperparameters

````python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
from tensorflow.keras.preprocessing.sequence import pad_sequences


# Hyperparameters
num_words = 10000   # Only top 10k most frequent words
maxlen = 200        # maximum lenght of each review

# 2. Cargar datos
(x_train, y_train), (x_test, y_test) = datasets.imdb.load_data(num_words=num_words)

print(len(x_train), len(x_test))   # 25000, 25000
print(x_train[0][:10])
````

Fill each sequence, so they all have the maximum lenght

````python
x_train_padded = pad_sequences(x_train, maxlen=maxlen)
x_test_padded  = pad_sequences(x_test, maxlen=maxlen)

print(x_train_padded.shape)  # (25000, 200)
print(x_test_padded.shape)   # (25000, 200)
````

````python
embedding_dim = 64

model = models.Sequential([
    layers.Embedding(input_dim=num_words,
                     output_dim=embedding_dim),
    layers.LSTM(64),
    layers.Dense(1, activation='sigmoid')   # binary target
])
````

````python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
````

````python
history = model.fit(
    x_train_padded,
    y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.2,
    verbose=2
)
````

````python
test_loss, test_acc = model.evaluate(x_test_padded, y_test, verbose=0)
print(f"Test accuracy: {test_acc:.4f}")
````
